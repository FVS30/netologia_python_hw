## Задание 1
Дана переменная, в которой хранится словарь, содержащий гео-метки для каждого пользователя (пример структуры данных приведен ниже). Вам необходимо написать программу, которая выведет на экран множество уникальных гео-меток всех пользователей.

```python
ids = {
'user1': [213, 213, 213, 15, 213],
'user2': [54, 54, 119, 119, 119],
'user3': [213, 98, 98, 35]
}
```

```python
uniq_geomet = set() # создаем множество пустое, чтобы туда потом добавить значения

for geo_list in ids.values(): #выводим списки каждого ключа
    for geomet in geo_list: # выводим все значения из списков
      uniq_geomet.add(geomet) # добавляем полученые значения из списков во множество чтобы получить только уникальные геометки
print(uniq_geomet)  # принтуем
```

```python
# Dict comprehension
uniq_geomet = {geomet for geo_list in ids.values() for geomet in geo_list}
print(uniq_geomet)
```

## Задание 2
Дана переменная, в которой хранится список поисковых запросов пользователя (пример структуры данных приведен ниже, но запросы потенциально могут быть любые). Вам необходимо написать программу, которая выведет на экран распределение количества слов в запросах в требуемом виде.

```python
queries = [
    'смотреть сериалы онлайн',
    'новости спорта',
    'афиша кино',
    'курс доллара',
    'сериалы этим летом',
    'курс по питону',
    'сериалы про спорт',
]
```

```python
counts = {} # хранилище словарь
queries_total = len(queries) # считаем количество запросов
for string in queries: # цикл в к-ом в каждой строке считается количество символов
    x = len(string.split())
    counts[x] = counts.get(x, 0) + 1 # здесь мы помещаем  количество слов в строке в key, а количество таких строк в value и так считает цикл, накапливая значения
    #print(counts)
for len_word, count in counts.items(): # метод .items() позволяет нам получить как key так и value | len_word - ключ где количество слов в строке | count - значение количества таких строк по количеству слов в строке
#разделение словаря на отдельные key|value
    percent = (count / queries_total) * 100
    #print(count)
    #print(len_word)
    print(f'Поисковых запросов, содержащих {len_word} слов(а): {percent:.2f}%')
```

## Задание 4
Дана переменная, в которой хранится статистика рекламных каналов по объемам продаж (пример структуры данных приведен ниже). Напишите программу, которая возвращает название канала с максимальным объемом продаж.

```python
stats = {'facebook': 55, 'yandex': 115, 'vk': 120, 'google': 99, 'email': 42, 'ok': 98}
```

```python
max_sales = 0
best_company = ''

for company, sales_volume in stats.items():
    if sales_volume > max_sales:
        max_sales = sales_volume
        best_company = company
print(f'Максимальный объем продаж на рекламном канале: {best_company}')

```

```python
# Альт max stats по ключам, через сравнение value
best_company = max(stats, key=stats.get)
print(f'Максимальный объем продаж на рекламном канале: {best_company}')
```

## Задание 5 (необязательно)
Дан список произвольной длины. Необходимо написать код, который на основе исходного списка составит словарь такого уровня вложенности, какова длина исхондого списка.

```python
my_list1 = ['2018-01-01', 'yandex', 'cpc', 100]
my_list2 = ['a', 'b', 'c', 'd', 'e', 'f']
```

```python
# Берем последний элемент как стартовое значение [-1] с конца списка
res = my_list1[-1]

# Идем по остальным элементам списка справа налево
# Срез [:-1] берет всё, кроме последнего. [::-1] разворачивает это. с шагом -1
for key in my_list1[:-1][::-1]:
    res = {key: res}

print(res)
```

```python
res = my_list2[-1]

for key in reversed(my_list2[:-1]):
    res = {key: res}

print(res)
```

```python
# проба через рекурсию
def req(my_list1):
    if len(my_list1) == 1:
        return my_list1[0]  # базовый случай: последний элемент

    first = my_list1[0]  # первый элемент — ключ
    rest = my_list1[1:]  # всё, кроме первого — остаток
    value = req(rest)  # строим значение из остатка
    return {first: value}


result = req(my_list1)
print(result)
```

## Задание 5 (необязательно)

Все ингредиенты для каждого блюда указаны в расчёте на одну порцию. Мы организуем застолье, и нам необходимо составить и вывести на экран список продуктов для покупки исходя из имеющихся рецептов. Количество людей, на которых мы будем готовить, мы по-прежнему будем запрашивать у пользователя. А также добавим функцию запроса ввода только нужных блюд. При этом мы никак не ограничиваем себя в том, что на каждого человека можно готовить несколько порций одного и того же блюда.

Пока реализуем две функции: запроса данных — количества гостей и списка блюд и формирования словаря для списка покупок аналогично домашнему заданию. Выведем результирующий словарь путём корректного вызова наших функций.

```python
cook_book = {
    'салат': [
        {'ingridient_name': 'сыр', 'quantity': 50, 'measure': 'гр'},
        {'ingridient_name': 'томаты', 'quantity': 2, 'measure': 'шт'},
        {'ingridient_name': 'огурцы', 'quantity': 20, 'measure': 'гр'},
        {'ingridient_name': 'маслины', 'quantity': 10, 'measure': 'гр'},
        {'ingridient_name': 'оливковое масло', 'quantity': 20, 'measure': 'мл'},
        {'ingridient_name': 'салат', 'quantity': 10, 'measure': 'гр'},
        {'ingridient_name': 'перец', 'quantity': 20, 'measure': 'гр'}
    ],
    'пицца': [
        {'ingridient_name': 'сыр', 'quantity': 20, 'measure': 'гр'},
        {'ingridient_name': 'колбаса', 'quantity': 30, 'measure': 'гр'},
        {'ingridient_name': 'бекон', 'quantity': 30, 'measure': 'гр'},
        {'ingridient_name': 'оливки', 'quantity': 10, 'measure': 'гр'},
        {'ingridient_name': 'томаты', 'quantity': 20, 'measure': 'гр'},
        {'ingridient_name': 'тесто', 'quantity': 100, 'measure': 'гр'},
    ],
    'лимонад': [
        {'ingridient_name': 'лимон', 'quantity': 1, 'measure': 'шт'},
        {'ingridient_name': 'вода', 'quantity': 200, 'measure': 'мл'},
        {'ingridient_name': 'сахар', 'quantity': 10, 'measure': 'гр'},
        {'ingridient_name': 'лайм', 'quantity': 20, 'measure': 'гр'},
    ]
}
```

```python
def get_info():
    dishes = input('Введите блюда через запятую')
    persons = int(input('Введите количество гостей'))
    return (dishes.split(','), persons)


def get_shop_list_by_dishes(dishes, person_count):
    shop_dict = {}
    for dish in dishes:
        for ingr in cook_book[dish]:
            key = (ingr['ingridient_name'], ingr['measure'])
            if key not in shop_dict:
                shop_dict[key] = ingr['quantity'] * person_count
            else:
                shop_dict[key] += ingr['quantity'] * person_count
    return shop_dict


dishes, persons = get_info()
print(get_shop_list_by_dishes(dishes, persons))
```
