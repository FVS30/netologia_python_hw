```python
import json # импорт модуля для работы с json
```

```python
 # создаем пустой словарь для работы в памяти
purchases = {}

# путь указывал через r тк делал в ide и там файлы в разных местах.
# читаем файл через контекстное меню. без encoding='utf-8' кирилица не отображалась
with open(r'/Вебинары/purchase_log.txt', 'r',
        encoding='utf-8'
) as f:
    # проходим циклом(читаем) файл по каждой строке
    for line in f:

        # преобразуем строку из json в словарь
        dict_ = json.loads(line)
        #print(dict_, type(dict_)) # делал проверку как в лекциях

        # извлекал значения по ключам
        user_id = dict_['user_id']
        category = dict_['category']

        # присвоил категории - value
        purchases[user_id] = category

# вывел выборку из первых пяти эл-тов
print(list(purchases.items())[:5])  # не дает выдать весь словарь не хватает памяти
```

## Задание № 2

```python
# открыл visit_log на чтение
with open(r'/Вебинары/visit_log.csv', 'r',
        encoding='utf-8'
) as v:
    # открыл funnel на запись(создал пустой файл)
    with open(r'funnel.csv', 'w', encoding='utf-8') as f:

        # проходим циклом(читаем) файл по каждой строке файла visit_log
        for line in v:

            # убираем пробельные символы и переносы strip()
            # потом разделяем через запятую split() и собираем в список
            line_list = line.strip().split(',')

            # извлекаем ID и источники
            user_id = line_list[0]
            source = line_list[1]

            # Ищем категорию покупки в нашем словаре 'purchases' по ID пользователя
            # метод .get() сейвовый не выдаст ошибку
            category = purchases.get(user_id)

            # Если категория найдена т.е. purchases.get вернул не None
            if category:

                # формируем список из эл-тов
                line_save = [user_id, source, category]
                # собираем список обратно в строку через запятую
                result = ','.join(line_save) + '\n'

                # Записываем сформированную строку в итоговый файл
                f.write(result)
```

```python
# проверяем вывод результата читаем файл funnel
with open('funnel.csv', 'r', encoding='utf-8') as f:

    for i, line in enumerate(f):

        print(line.strip())
        if i >= 10:  # Выведем только первые 10 строк и сразу стопим, чтобы не засорять экран
            break
```
