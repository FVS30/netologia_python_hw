```python
# coding: utf-8
import requests

class Rate:
    def __init__(self, format='value', diff=False):
        self.format = format
        self.diff = diff

    def exchange_rates(self):
        """
        Возвращает ответ сервиса с информацией о валютах в виде:

        {
            'AMD': {
                'CharCode': 'AMD',
                'ID': 'R01060',
                'Name': 'Армянских драмов',
                'Nominal': 100,
                'NumCode': '051',
                'Previous': 14.103,
                'Value': 14.0879
                },
            ...
        }
        """
        r = requests.get('https://www.cbr-xml-daily.ru/daily_json.js')
        return r.json()['Valute']

    def make_format(self, currency):
        """
        Возвращает информацию о валюте currency в двух вариантах:
        - полная информация о валюте при self.format = 'full':
        Rate('full').make_format('EUR')
        {
            'CharCode': 'EUR',
            'ID': 'R01239',
            'Name': 'Евро',
            'Nominal': 1,
            'NumCode': '978',
            'Previous': 79.6765,
            'Value': 79.4966
        }

        Rate('value').make_format('EUR')
        79.4966
        """
        response = self.exchange_rates()

        if currency in response:  # проверка кода валюты
            # № 1 выводит все данные о валюте
            if self.format == 'full':
                return response[currency]

            # № 2 выводит только курс валюты
            if self.format == 'value':

                # проверка режима разницы курсов
                if self.diff == True:
                    # Вычитаем из текущего значения (Value) прошлое (Previous)
                    return response[currency]['Value'] - response[currency]['Previous']
                else:
                    # Если diff выключен (False) — просто отдаем текущий курс
                    return response[currency]['Value']

        return 'Error'

    def eur(self):
        """Возвращает курс евро на сегодня в формате self.format"""
        return self.make_format('EUR')

    def usd(self):
        """Возвращает курс доллара на сегодня в формате self.format"""
        return self.make_format('USD')

    def AZN(self):
        """Возвращает курс азербайджанского маната на сегодня в формате self.format"""
        return self.make_format('AZN')

    # Здесь создаю функцию согласно заданию 1
    def name_max_value_exchange_rate(self):

        """
        Метод для поиска валюты с максимальным значением курса.
        Возвращает кортеж: (Название валюты, Курс)
        """

        # Задаем переменные для хранения рез-тов
        name_currency = None  # это название самой дорогой переменной
        max_rate = 0  # Это максимальное значение курса

        # Получаем актуальные данные о курсах, вызывая метод родителя
        response = self.exchange_rates()

        # Перебираем словарь валют (name_valut — код валюты/ключ, data — словарь с деталями/значения)
        for name_valut, data in response.items():

            # Если текущее значение 'Value' больше того, что мы уже видели:
            if data['Value'] > max_rate:
                # Обновляем наш "максимум" курса
                max_rate = data['Value']
                # Указываем название из словаря с ключом ['Name'] /Американский доллар США/, а не USD
                name_currency = data['Name']

        # Возвращаем результат в виде пары значений Название, курс
        return name_currency, max_rate
```

```python
class Employee:
    def __init__(self, name, seniority):
        self.name = name
        self.seniority = seniority

        self.grade = 1   # стартовый грейд

    def grade_up(self):
        """Повышает уровень сотрудника"""
        self.grade += 1

    def publish_grade(self):
        """Публикация результатов аккредитации сотрудников"""
        print(self.name, self.grade)

    def check_if_it_is_time_for_upgrade(self):
        pass
```

```python
class Designer(Employee):
    def __init__(self, name, seniority, int_awards=2):
        super().__init__(name, seniority=(int_awards * 2))

    def check_upgrade_at_designer(self):
        """
        Данный метод должен проверять повышение дизайнеров при достижении им 7 баллов рейтинга.
        """

        self.seniority += 2  # здесь у нас за одну премию получаем 2 бала

        # проверяем, каждые 7 баллов = 1 грейд и повышаем соответственно
        # тк у нас повышается по 2 балла мы можем упустить возможность повышения,
        # поэтому мы сравниваем разницу между текущим повышением и прошлым(на 2 балла ниже)
        if (self.seniority // 7) > ((self.seniority - 2) // 7):
            self.grade_up()

        # возвращаем результат
        return self.publish_grade()
```

```python
den = Designer('Денис', 0)
```

```python
for i in range(20):
    den.check_upgrade_at_designer()
```
