```python
class Task_1(Rate):  # создаю дочерний класс из род.класса Rate

    def __init__(self, format='value', diff=True):
        # Вызываем настройщик родительского класса Rate через super()
        # Это нужно, чтобы у Task_1 появилось свойство self.format
        super().__init__(format)
        self.diff = diff

    def name_max_value_exchange_rate(self):

        """
        Метод для поиска валюты с максимальным значением курса.
        Возвращает кортеж: (Название валюты, Курс)
        """

        # Задаем переменные для хранения рез-тов
        name_currency = None  # это название самой дорогой переменной
        max_rate = 0  # Это максимальное значение курса

        # Получаем актуальные данные о курсах, вызывая метод родителя
        response = self.exchange_rates()

        # Перебираем словарь валют (name_valut — код валюты/ключ, data — словарь с деталями/значения)
        for name_valut, data in response.items():

            # Если текущее значение 'Value' больше того, что мы уже видели:
            if data['Value'] > max_rate:
                # Обновляем наш "максимум" курса
                max_rate = data['Value']
                # Указываем название из словаря с ключом ['Name'] /Американский доллар США/, а не USD
                name_currency = data['Name']

        # Возвращаем результат в виде пары значений Название, курс
        return name_currency, max_rate
```

```python
my_task = Task_1(format='full')
```
