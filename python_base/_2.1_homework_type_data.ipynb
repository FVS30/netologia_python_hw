## Задание № 1
Дана переменная, в которой хранится слово из латинских букв. Напишите код, который выводит на экран:

- среднюю букву, если число букв в слове нечетное;

- две средних буквы, если число букв четное.

```python
word = input('Введите слово: ')    # Вводим слово

if len(word) % 2 == 0:    # Проверка четности, делится на 2 без остатка - "четное" иначе "нечентное"
    mid = len(word) // 2
    print(word[mid - 1] + word[mid])

else:
    mid = len(word) // 2
    print(word[mid])
```

```python
# альт
word = input('Введите слово: ')    # Вводим слово (hello и test)
# вывести символы из переменной word с индексами , где индексы расчитываются по формату начало диапозона и конец, в зависимости от финкции длинна слова в символах
# Пример : [2:3] → только 1 "l" символ/ [1:3] два символа "e" и "s"
print(word[(len(word)-1)//2:(len(word)+2)//2])
```

## Задание № 2
Напишите программу, которая последовательно запрашивает у пользователя числа (по одному за раз) и после первого нуля выводит сумму всех ранее введенных чисел.

```python
total_sum = 0                                    # старт значение

y = int(input('Введите цифру: '))                # вводим цифры для работы цикла

while y != 0:                                    # цикл
    total_sum += y                               # формула расчета по которой действует цикл
    y = int(input('Введите слудующую цифру: '))  # вводим слуд цифру

print('Результат :', total_sum)                  # результат
```

```python
# альт решение через моржовый оператор для оптимизации
a = 0

# "Моржовый оператор" :=  оператор возвращает False только при 0, также введенное число сразу попадает в переменную number, если !=0
while number := int(input("введите число: ")):
    a += number
print(a)
```

## Задание № 5

Имеется список с транспортными номерами. Необходимо написать код, который проверит каждый номер на валидность (1 буква, 3 цифры, 2 буквы, 2-3 цифры).

Обратите внимание, что не все буквы кириллического алфавита используются в транспортных номерах.

Если номер валиден, то вывести его в нужном формавте (пример ниже), а если не валиден — вывести текст.
При решении помогут регулярные выражения, с которыми будет знакомство на практике.

```python
# План решения:
# 1. подключить регулярки import re
# 2. составить шаблон поиска из 2 групп
# 3. Если номер совпадает по правилам из шаблона то номер валиден и нужно привести к форме вывода разделенной на две группы
# 4. 1 группа - номер 2 группа-регион
# 5 если не подходит по шаблон не валиден
```

```python
# Примеп:
car_ids = [
    'А222ВС96',
    'АБ22ВВ193',
    'А222ВС102',
    'В456ОР777',
    'К789НМ123',
    'М123ТХ45',
    'Н321УХ999',
    'АБ22ВВ193',
    'С12РТ67',
    'ОП456РТ78',
    'Т999КК9',
    'Х4567ММ22'
]
print(type(car_ids))
```

```python
import re

# В задании возникли сложности пришлось погуглить, тк не получалось сделать через re.search и  re.findall
```

```python
pattern = r"([АВЕКМНОРСТУХ]\d{3}[АВЕКМНОРСТУХ]{2})(\d{2,3})"

for carid in car_ids:
    fnomber = re.fullmatch(pattern, carid)                  # загуглил, что  регулярка фуллмач для целого списка, в отличии от re.search и  re.findall
    if fnomber:
        full_number = fnomber.group(0)
        number = fnomber.group(1)
        region = fnomber.group(2)
        print(f'Номер {number} валиден. Регион: {region}')

    else:
        print(f'Номер {carid} не валиден')
```

## Задание № 6
Дан поток логов по количеству просмотренных страниц для каждого пользователя (пользователь,дата;просмотры).
Вам необходимо написать алгоритм, который считает среднее значение просмотров на пользователя. Т. е. надо посчитать отношение суммы всех просмотров к количеству уникальных пользователей. И тут регулярные выражения облегчат немного реализацию.

```python
# План решения:
# 1. подключить регулярки import re
# 2. составить шаблон поиска: из нескольких групп
# 3. по паттернам составить два списка users(сумма уникальных пользователей (убрать дубли "user4 user4 user3 => user4 user3")) и views (просто сумма просмотров через сумму в списке)
# 4 расчитать ср.знач. просмотров =  округлив до 2 знаков views/users округлив до 2 знаков
```

```python
# пример:
stream = [
    'user4,2021-01-01;3',
    'user3,2022-01-07;4',
    'user2,2022-03-29;1',
    'user1,2020-04-04;13',
    'user2,2022-01-05;7',
    'user1,2021-06-14;4',
    'user3,2022-07-02;10',
    'user4,2021-03-21;19',
    'user4,2022-03-22;4',
    'user4,2022-04-22;8',
    'user4,2021-05-03;9',
    'user4,2022-05-11;11'
]
```

```python
import re
```

```python
pattern_groups = r'(user\d+),.*?;(\d+)'

# Создаём пустые списки
users_list = []
views_list = []

# цикл
for i in stream:
    x = re.search(pattern_groups, i)       # Ищем по патерну в каждой строке
    user, views_str = x.groups()           # Мн. присваение
    views = int(views_str)                 # перевод в тип д. int

    users_list.append(user)                # наполняем списки
    views_list.append(views)

total_views = sum(views_list)              # суммир. все числа просмотров в списке
unique_users = len(set(users_list))        # вычисляем кол-во символов в списке, преобразует во множество, убрать дубли

avg_veiw = round((total_views / unique_users), 2)

# для проверки данных
#print(f'количество просмотров: {total_views}')
#print(f'уникального пользователя: {unique_users}')

print(f'Среднее количество просмотров на уникального пользователя: {avg_veiw:.2f}')
```

```python
# Альт вариант
pattern_groups = r'(user\d+),.*?;(\d+)'

total_veiw = 0
unique_users = set()                          # для того, чтобы через set выдало множества,

for i in stream:
    match_ = re.search(pattern_groups, i)
    user, views_str = match_.groups()         # мн.присвоение
    views = int(views_str)                    # переводим в int

    total_veiw += views
    unique_users.add(user)

avg_veiw = round(total_veiw / len(unique_users), 2)

# для проверки данных
#print(f'количество просмотров: {total_views}')
#print(f'уникального пользователя: {unique_users}')

print(f'Среднее количество просмотров на уникального пользователя: {avg_veiw:.2f}')
```
