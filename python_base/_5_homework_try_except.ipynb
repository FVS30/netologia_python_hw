## Задание 2


Дан поток дат в формате YYYY-MM-DD, в которых встречаются некорректные значения:


```python
stream = ['2018-04-02', '2018-02-29', '2018-19-02']  # наш список, умные кавычки в ручную поправил
```

Напишите функцию, которая проверяет эти даты на корректность.

То есть для каждой даты возвращает True — дата корректна или False — некорректная.

```python
# Создаем ф-цию, которая будет проверять строку с датой на соответствие шаблону с выводом логич. выражения
def chec_date(date_str: str) -> bool:

    """
    Ф-ция к-ая должна проверять строку даты на соответстввие шаблону.
    Результат ф-ции должен вернуть логическое соотв-вие или НЕ соотв-вие
    """

    try:
        datetime.strptime(date_str, '%Y-%m-%d')
        return True

    except (SyntaxError, ValueError):
        return False
```

```python
# мы пробегаемся по датам из списка stream и каждый элемент списка проверяем функцией chec_date
for date in stream:

    result = chec_date(date)
    print(f'Дата: {date} прошла проверку? - {result}')

    # Если строка выявлено несоответствие будет выведен доп комментарий
    if not result:
        print(f'Формат даты должен быть YYYY-MM-DD или дата отсутствует')
```

## Задание 3
Напишите функцию date_range, которая возвращает список дат за период от start_date до end_date.

Даты должны вводиться в формате YYYY-MM-DD.

В случае неверного формата или при start_date > end_date должен возвращаться пустой список.

```python
def date_range(start_date: str, end_date: str) -> list:
    """
    Возвращает список дат за период от start_date до end_date (не включительно).
    В случае неверного формата или start_date > end_date возвращает пустой список.

    На входе два параметра начальная дата и конечная дата в формате строк
    На выходе список дат
    """

    try:
        # Преобразуем строки в объекты datetime через метод strptime
        start_dt = datetime.strptime(start_date, '%Y-%m-%d')
        end_dt = datetime.strptime(end_date, '%Y-%m-%d')

        # Если start_date > end_date — возвращаем пустой список
        if start_dt > end_dt:
            return []

        # создаем пустой список для наполнения
        result = []
        # вводим аргумент текущей даты
        current_date = start_dt

        # Проходим по дням от start_dt до end_dt (не включая end_dt знак <)
        while current_date < end_dt:
            result.append(current_date.strftime('%Y-%m-%d'))  # добавляем в список строку с датой
            # счетчик увеличивается на 1 день с прохождением цикла
            current_date += timedelta(days=1)

        return result

    except ValueError:
        return []  # если формат даты некорректный
```

```python
start_date = '2026-04-29'
end_date = '2026-05-05'
```

```python
date_x = date_range(start_date, end_date)
print(date_x)
```
