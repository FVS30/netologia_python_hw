## Задание 1

### Дано:

```python
# Перечень всех книг:
books = [
    {
        'genre': 'поэзия', 'number': '978-5-1000-1234-7', 'title': 'Евгений Онегин',
        'author': 'Александр Пушкин'
    },
    {
        'genre': 'фэнтези', 'number': '88006', 'title': 'Властелин колец',
        'author': 'Джон Р. Р. Толкин'
    },
    {
        'genre': 'детектив', 'number': 'D-1122', 'title': 'Безмолвный свидетель',
        'author': 'Агата Кристи'
    }
]
```

```python
# Перечень полок, на которых хранятся книги (если книга есть в books, то она обязательно должна быть и в directories)
directories = {
    '1': ['978-5-1000-1234-7', '88006'],
    '2': ['D-1122'],
    '3': []
}
```

### Условия выполнения:

Пункт 1. Пользователь по команде "book_info" может найти название книги и её автора

Пункт 2. Пользователь по команде "shelf" может по названию книги узнать, на какой полке она хранится

Пункт 3. Пользователь по команде "all" может увидеть полную информацию по всем книгам

Пункт 4. Пользователь по команде "add_shelf" может добавить новую полку

Пункт 5. Пользователь по команде "del_shelf" может удалить существующую полку из данных (только если она пустая)


```python
def book_info(book_number):
    """
    Ф-ция для поиска названия и автора книги по номеру
    """

    # В цикле ищем книгу в списке books
    for book in books:

        # Проверка соответствия номера книги в цикле с номером от пользователя
        if book['number'] == book_number:

            # Возврат при совпадении
            return f"\n Название книги: {book['title']} Автор: {book['author']}"

    # Если ничего не нашли, возвращаем ошибку
    return f"Книга не найдена в базе"
```

```python
def get_shelf_by_number(book_number):
    """
    Вспомогательная ф-ция № 1 для поиска номера полки по номеру книги.
    Возвращ-ет None при отсут-вии данных
    """

    # Цикл для получения номеров полок - shelf_number и перечня книг н полке - book_list
    for shelf_number, book_list in directories.items():

        # Проверка есть ли номер книги на полке, если да - возврат номера полки
        if book_number in book_list:
            return shelf_number

    # Если ничего не нашли, возвращаем ошибку
    return None
```

```python
def get_number_book_by_title(title):
    """
    Вспом-ая ф-ция № 2 для получения номера книги по её названию.
    Т.к. в 'directories' хранятся только номера,
    чтобы найти путь к полке через название.
    """

    # Перебираем список, чтобы найти искомое название книги
    for book in books:

        # если совпало с пользовательским вводом выводим номер книги
        if book['title'] == title:
            return book['number']

    # Если ничего не нашли, возвращаем ошибку
    return None
```

```python
def shelf(title):
    """
    Ф-ция, работает с помощью ф-ций № 1 и № 2:
    находит номер полки, используя только название книги.
    Связывает воедино поиск ID(номеру) по названию и поиск полки по ID.
    """

    # превращаем название книги в ID книги
    target = get_number_book_by_title(title)
    # проверяем нашли ли такое название, если ID не None
    if target is not None:

        # возвращаем ф-цию, которая покажет нам на какой полке лежит книга
        return get_shelf_by_number(target)

    # Если ничего не нашли, возвращаем ошибку
    return f'Книга не найдена в базе'
```

```python
def get_all_books_info():
    """
    Ф-ция для формирования полного отчета о всех книгах в библиотеке.
    Собирает данные из 'books' и 'directories' и объединяет их в одну строку.
    """

    # Создаем пустой список, куда будем складывать готовые текстовые строки для каждой книги
    book_in_all_info = []

    # цикл, к-ый проходит по каждой книге(dict) в списке books
    for book in books:
        #извлекаем методом .get() значения по ключам и присваиваем соотв-щей переменной
        number = book.get('number')
        genre = book.get('genre')
        title = book.get('title')
        author = book.get('author')

        # вызываем вспомогательную ф-цию, чтобы получить информацию на какой полке находится книга с ID(номером)
        shelf_number = get_shelf_by_number(number)

        # собираем в f-cnhjre
        item = f'\n №: {number}, жанр:{genre}, название:{title}, автор:{author}, номер полки:{shelf_number}'
        # добавляем в ранее созданный пустой список
        book_in_all_info.append(item)

    # склеиваем все строки из списка в одну большую строку(перечень) через .join()
    return '\n'.join(book_in_all_info)
```

```python
def add_shelf(shelf_number):
    """
    Функция для добавления новой пустой полки в библиотеку.
    Принимает номер полки, проверяет её наличие и создает запись в 'directories'.
    """

    # Создаем перечень со всеми текущими полками через ключи(номера полок)
    all_shelves = ", ".join(directories.keys())

    # Проверка на существование полки перед созданием. с соотв-щим оповещением
    if shelf_number in directories:
        return f'Такая полка уже существует. Текущий перечень полок: {all_shelves}'

    # Создаем новую полку
    # метод setdefault(key, default) безопасно создает полку,
    # если нет ключа(key) создает его вместе с default значением - [](пустого словаря)
    directories.setdefault(shelf_number, [])

    # Обновляем перечень со всеми текущими полками (после создания новых)
    all_shelves = ", ".join(directories.keys())

    #Инф-е оповещение о создании полки и выводе всех полок на экран
    return f'Полка добавлена. Текущий перечень полок: {all_shelves}'
```

```python
def del_shelf(shelf_number):
    """
    Ф-ция для удаления пустой полки.
    Защищает от случайного удаления полок, на которых еще стоят книги.
    """

    # Подготавливаем перечень со всеми текущими полками
    all_shelves = ", ".join(directories.keys())

    # Проверяем есть ли полка с номером, к-ый ввел user, если нет выдаем ошибку с перечнем полок
    if shelf_number not in directories:
        return f'Ошибка такой полки нет {all_shelves}'

    else:

        # Проверяем есть ли книги на полке. Если len = 0 -> пустая полка
        if len(directories[shelf_number]) == 0:
            # После проверки удаляем полку(key) безвозвратно
            del(directories[shelf_number])
            # Пересобираем перечень полок после удаления
            all_shelves = ", ".join(directories.keys())
            # Инф-е оповещение об удалении полки и выводе всех полок на экран
            return f'Полка удалена. Текущий перечень полок: {all_shelves}.'

        else:

            # Если проверка не пройдена, собираем перечень полок и возвращаем соотв-щий ответ
            all_shelves = ", ".join(directories.keys())
            return f'На полке есть книги, удалите их перед удалением полки. Текущий перечень полок: {all_shelves}.'
```

## Задание 2 (необязательное)

### Условия выполнения:
Вам необходимо дополнить программу из задания 1 более продвинутыми командами

Пункт 1. Пользователь по команде "add_book" может добавить новую книгу в данные

Пункт 2. Пользователь по команде "del_book" может удалить книгу из данных

Пункт 3. Пользователь по команде "move" может переместить книгу с полки на полку




```python
def add_book(number, genre, title, author, shelf_number):
    """
    Ф-ция для регистрации новой книги.
    Одновременно создает запись в общем реестре (books) и
    ставит книгу на конкретную полку (directories).
    """

    # Проверка № 1. Существует ли полка, на к-ую мы хотим положить книгу?
    # Если полки нет, мы нет, то ее нужно создать соотв-щей командой 'add_shelf'
    if shelf_number not in directories:
        return f'\n Такой полки не существует. Добавьте полку командой add_shelf.'

    # Проверка № 2: Уникальность номера книги.
    # Пробегаем по всем книгам в списке books, чтобы убедиться, что такого ID(номера) еще нет.
    for book in books:
        if book['number'] == number:
            return f'Книга с таким №: {number} уже существует базе'

    # Создаем новый словарь, здесь вся инфа о книге, кроме номера полки
    new_book = {
        'number': number,
        'genre': genre,
        'title': title,
        'author': author,
    }

    # Обновляем БД:
    books.append(new_book)  # добавление в общий список books
    directories[shelf_number].append(number
    )  # связываем(добавляем) ID книги на полку в словаре directories

    return f'\n Книга добавлена. Текущий список книг: {get_all_books_info()}:'
```

```python
def del_book(number):
    """
    Функция для полного удаления книги из системы.
    Удаляет описание из 'books' и номер из 'directories'.
    """

    # Создаем переменную, как индикатор, пока там пусто.
    found_book = None
    # Перебор книг в списке книг
    for book in books:
        # Запоминаем словарь с искомой книгой, если номера (ID) сошлись
        if book['number'] == number:
            found_book = book
            break  # как шали - сразу выходим из цикла

    # Если ID = None выдаем ошибку с перечнем всех книг, к-ый вызывается ф-цией
    if found_book is None:
        return f'\n Книга не найдена в базе.Текущий список книг:{get_all_books_info()}'

    # Если книга была найдена, удаляем ее из общего списка books
    books.remove(found_book)

    # Удаление ID книги с полки в directories.
    # Метод .values() перебирает только значения словаря
    for book_list in directories.values():
        # Если такой ID есть на полке(в списке) удаляем и сразу останавливем цикл
        if number in book_list:
            book_list.remove(number)
            break

    # Отчет об удалении книги с текущ. списком книг
    return f'\n Книга удалена. Текущий список книг:{get_all_books_info()}'
```

```python
def move(number, new_shelf):
    """
    Функция для перемещения книги с одной полки на другую.
    Проверяет существование новой полки, находит текущую полку книги и меняет её местоположение.
    """

    # Собираем перечень всех книг
    all_shelves = ", ".join(directories.keys())

    # Проверка № 1: Существует ли полка, на которую мы хотим переставить книгу?
    if new_shelf not in directories:
        return f'Такой полки не существует. Текущий перечень полок: {all_shelves}.'

    # Начинаем поиск книги по всем полкам в словаре directories
    # current_shelf — номер полки (ключ), book_list_on_shelf — список книг на ней (значение)
    for current_shelf, book_list_on_shelf in directories.items():
        # Если номер книги нашелся в списке текущей полки
        if number in book_list_on_shelf:

            # Проверка № 2: Стоит ли книга на той полке, куда ее хочет переместить user?
            if current_shelf == new_shelf:
                return f'Книга уже на этой полке'

            else:
                # Удаляем номер книги из списка(полки). Метод remove() оставит ее в памяти для перемещения
                book_list_on_shelf.remove(number)

                # Обращаемся к новой полке по ключу и добавляем туда этот номер
                directories[new_shelf].append(number)
            # прерываем как только нашли и переместили. и вернули ин-фу о том что книга перемещена и выдаем отчет по книгам
            return f'\n Книга перемещена. Текущий список книг:{get_all_books_info()}'

    # Если такой книги нет. Не пройдена проверка 'if number in book_list_on_shelf' возвращаем ответ о том что книга не найдена в бд
    return f'\n Книга не найдена в базе. Текущий список книг:{get_all_books_info()}'
```

```python
def main():
    """
    Агрегатор ф-ций(команд) для работы с библиотекой
    ## Ф-ция будет работать пока ее не остановить командой 'q' ##

    Перечень команд:
    'book_info' - найти название книги и её автора
    'shelf' - названию книги узнать, на какой полке она хранится
    'get_all_books' - увидеть полную информацию по всем книгам. all дает ошибку
    'add_shelf' - может добавить новую полку
    'del_shelf' - может удалить существующую полку из данных (только если она пустая)
    'add_book' - может добавить новую книгу в данные
    'del_book' - может удалить книгу из данных
    'move' - перемещение книги между полок
    'q' - завершение
    """
    while True:
        command = input('Введите команду:')

        match command:
            case 'q':
                print('Работа программы завершена')
                break

            case 'book_info':
                # вводится пользовательский номер книги
                user_search_number = input('Введите номер книги:')
                # присваивается результат функции
                result = book_info(user_search_number)
                # вывод результата

                print(result)

            case 'shelf':
                user_search_title = input('Введите название книги:')
                result = shelf(user_search_title)

                print(result)

            case 'get_all_books':
                print(get_all_books_info())

            case 'add_shelf':
                shelf_add = input('Введите номер полки:')
                result = add_shelf(shelf_add)

                print(result)

            case 'del_shelf':
                delete_shelf = input('Введите номер полки:')
                result = del_shelf(delete_shelf)

                print(result)

            # допы:
            case 'add_book':
                # нужна ли здесь проверка на номер книги? для user удобства, чтобы выдать ошибку по отсутствию номер?
                number_book = input('Введите номер книги:')
                genre_book = input('Введите жанр книги')
                title_book = input('Введите название книги:')
                author_book = input('Введите автора книги:')
                shelf_book = input('Введите полку для хранения:')

                result = add_book(number_book, genre_book, title_book, author_book, shelf_book)

                print(result)

            case 'del_book':
                user_number_delete = input('Введите номер книги:')
                result = del_book(user_number_delete)

                print(result)

            case 'move':
                number_book = input('Введите номер книги:')
                new_shelf_number = input('Введите номер полки:')

                print(move(number_book, new_shelf_number))
```
