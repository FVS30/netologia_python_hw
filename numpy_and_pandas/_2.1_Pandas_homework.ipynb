### Задание № 1:
Датасет для домашнего задания находится в материалах занятия - файлы для домашнего задания «Библиотека Pandas».

Определите, какому фильму было выставлено больше всего оценок 5.0.

```python
import pandas as pd
```

```python
# открываем csv-шник, выводим шапку на экран
data_rating = pd.read_csv('ratings.csv')
data_rating.head()
```

```python
# выводим оценки строго= 5
ratings_5 = data_rating[data_rating['rating'] == 5]
ratings_5.head()
```

```python
# value_counts подсчитываем количество значений по колонке фильмid
hg_rating = ratings_5['movieId'].value_counts()

# hg_rating.head() # вывод шапки в первых 5 фильмов для наглядности --топ 5

hg_rating  # вывод одного фильма топ 1
```

```python
print(f' Фильм, к-ому было выставлено больше всех 5.0 оценок \n movieId: {hg_rating.index[0]}, values: {hg_rating.values[0]}')
```

### Задание № 2:
По данным файла power.csv посчитайте суммарное потребление стран Прибалтики (Латвия, Литва и Эстония) категорий 4, 12 и 21 за период с 2005 по 2010 год. Не учитывайте в расчётах отрицательные значения quantity.

```python
data_power = pd.read_csv(
    r'C:\Users\fokin\Desktop\Перекатываемся\Нетология\Python\Вебинары\Pandas и numpy\power.csv'
)
data_power.head()
```

```python
# просто чтобы посмотреть есть ли в списке искомые страны Латвия — Latvia / Литва — Lithuania / Эстония — Estonia
data_power[data_power['country'].str.contains('ia', case=False)]['country'].unique()
```

```python
filter_country = data_power[
    (data_power['country'] == 'Latvia') | (data_power['country'] == 'Lithuania') | (
            data_power['country'] == 'Estonia')]

filter_country
```

```python
# после выполнения задумался об оптимизации, чтобы не писать грамоздкую конструкцию, если стран будет 25, 30,
# такое условие просто не написать, ИИ-шка подсказала использовать метод .isin() на всякий случай закомментрировал этот метод
# filter_country = data_power[data_power['country'].isin(['Latvia', 'Lithuania', 'Estonia'])]
# filter_country
```

```python
filter_category = filter_country[
    (filter_country['category'] == 4) | (filter_country['category'] == 12) | (
            filter_country['category'] == 21)]

filter_category
```

```python
# filter_category = filter_country[filter_country['category'].isin([4, 12, 21])]

# filter_category
```

```python
filter_ccy = filter_category[
    (filter_category['year'] >= 2005) & (filter_category['year'] <= 2010) & (
                filter_category['quantity'] >= 0)]
filter_ccy
```
