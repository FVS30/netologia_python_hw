### Задание № 2:

Используйте файл keywords.csv.

Нужно написать гео-классификатор, который каждой строке сможет выставить географическую принадлежность определённому региону. Т. е. если поисковый запрос содержит название города региона, то в столбце ‘region’ пишется название этого региона. Если поисковый запрос не содержит названия города, то ставим ‘undefined’.

Правила распределения по регионам Центр, Северо-Запад и Дальний Восток:

Результат классификации запишите в отдельный столбец region.

```python
geo_data = {
    'Центр': ['москва', 'тула', 'ярославль'],
    'Северо-Запад': ['петербург', 'псков', 'мурманск'],
    'Дальний Восток': ['владивосток', 'сахалин', 'хабаровск']
}
```

```python
df_key = pd.read_csv('keywords.csv')
df_key.head()
```

```python
def geo_classify(row):
    '''
    ф-ция выставляет географическую принадлежность к определенному региону из geo_data

    Примечание:
    ф-ция достатолчно простая и не распознает падежи и склонения
    '''

    # контейнер для ячейки, также отбросил разницу в регистрах
    geo_class = row['keyword'].lower()

    # пробегаемся по словарю geo_data и достаем из него каждое название города (влож цикл)
    for region, cities in geo_data.items():
        for city in cities:
            if city in geo_class:
                return region

    return 'undefined'
```

```python
df_test2['region'] = df_test2.apply(geo_classify, axis=1)
df_test2
```

### Задание № 3:
Есть мнение, что раньше снимали настоящее кино, не то что сейчас.

Ваша задача — проверить это утверждение, используя файлы с рейтингами фильмов из прошлого домашнего занятия:
файл movies.csv и ratings.csv из базы.
Нужно проверить, верно ли, что с ростом года выпуска фильма его средний рейтинг становится ниже.

Вы не будете затрагивать субьективные факторы выставления этих рейтингов, а пройдётесь по алгоритму:

1. В переменную years запишите список из всех годов с 1950 по 2010 года.

2. Напишите функцию production_year, которая каждой строке из названия фильма выставляет год выпуска. Не все названия фильмов содержат год выпуска в одинаковом формате, поэтому используйте алгоритм:

- для каждой строки пройдите по всем годам списка years;
- если номер года присутствует в названии фильма, то функция возвращает этот год, как год выпуска;
- если ни один из номеров года списка years не встретился в названии фильма, то возвращается 1900 год.
3. Запишите год выпуска фильма по алгоритму пункта 2 в новый столбец ‘year’.

4. Посчитайте средний рейтинг всех фильмов для каждого значения столбца ‘year’ и отсортируйте результат по убыванию рейтинга.

```python
years_list = list(range(1950,2011))
years_list
```

```python
df_movies = pd.read_csv('movies.csv')
df_movies.head()
```

```python
df_ratings = pd.read_csv('ratings.csv')
df_ratings.head()
```

```python
def production_year(row):
    '''
    ф-ция, которая каждой строке из названия фильма выставляет год выпуска 1/0
    '''

    str_year_release = row['title']

    for year in years_list:
        if '(' + str(year) + ')' in str_year_release:
            return year

    return 1900
```

```python
df_movies['year'] = df_movies.apply(production_year, axis=1)
df_movies
```

```python
joined_tables = df_ratings.merge(df_movies, on='movieId')
joined_tables
```

```python
# сортировка по убыванию рейтинга и высчитывание среднего значения по 'rating'
# joined_tables.sort_values('rating', ascending=False) # шаблон как пишется сортировка в отдельную строку

joined_tables.groupby('year')[['rating']].mean().sort_values('rating', ascending=False)
```
